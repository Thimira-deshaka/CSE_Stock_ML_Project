```python
import wbgapi as wb
import yfinance as yf
import pandas as pd
import os

# Create directory if it doesn't exist (failsafe)
os.makedirs('../data/raw', exist_ok=True)

# 1. Download Stock Data
ticker = "^CSE"
start_date = "2018-01-01"
end_date = "2024-12-31"

stock_data = yf.download(ticker, start=start_date, end=end_date)
# Saving to the RAW data folder
stock_data.to_csv("../data/raw/cse_historical_data.csv")

# 2. Download Macro Data
indicators = {'FP.CPI.TOTL.ZG': 'Inflation', 'NY.GDP.MKTP.KD.ZG': 'GDP_Growth'}
macro_data = wb.data.DataFrame(indicators.keys(), 'LKA', range(2018, 2025))
# Saving to the RAW data folder
macro_data.to_csv("../data/raw/srilanka_macro_data.csv")

print("Data collection complete. Files saved in data/raw/")
```
